```python
import pandas as pd

# Load data
weather = pd.read_csv(r"Data\Raw\nasa_power_1984_2016.csv")
rain = pd.read_csv(r"Data\Raw\customized_daily_rainfall_data.csv")

# Convert dates
weather["date"] = pd.to_datetime(weather["date"])

rain["date"] = pd.to_datetime(
    rain[["Year", "Month", "Day"]],
    errors="coerce"
)

# Merge
merged = weather.merge(
    rain[["Station", "date", "Rainfall"]],
    on=["Station", "date"],
    how="left"
)
merged.rename(columns={"Rainfall": "BMD_Rainfall"}, inplace=True)

merged.to_csv(r"Data\Merged\Merged.csv", index=False)
```
